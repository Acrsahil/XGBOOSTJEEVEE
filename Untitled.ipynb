```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error


# ==========================================
# LOAD DATA
# ==========================================

train_raw = pd.read_csv("Train_Data.csv")   # 2025
test_raw = pd.read_csv("Test_Data.csv")     # 2026

train_raw["order_date"] = pd.to_datetime(train_raw["order_date"])
test_raw["order_date"] = pd.to_datetime(test_raw["order_date"])

print("Train:", train_raw["order_date"].min(),
      "→", train_raw["order_date"].max())


print("Test:", test_raw["order_date"].min(),
      "→", test_raw["order_date"].max())
```

```python
# combining only for feature creation
all_data = pd.concat(
    [train_raw, test_raw],
    ignore_index=True
)

all_data = all_data.sort_values(
    ["product_id", "order_date"]
).reset_index(drop=True)
all_data['order_date']
```

```python
# Aggregate to daily sales

daily = (
    all_data
    .groupby(
        [
            "product_id",
            "order_date"
        ],
        as_index=False
    )
    .agg(
        sales_qty=("qty_invoiced", "sum"),
        price=("jeevee_variant_price", "mean"),
        product_name=("product_name", "first"),
        brand_name=("brand_name", "first"),
        web_category=("web_category", "first")
    )
)
daily.head()
```

```python
# Create Complete Dates

products = (
    daily[
        [
            "product_id",
            "product_name",
            "brand_name",
            "web_category"
        ]
    ]
    .drop_duplicates("product_id")
)

dates = pd.DataFrame({
    "order_date": pd.date_range(
        daily["order_date"].min(),
        daily["order_date"].max(),
        freq="D"
    )
})

products["_key"] = 1
dates["_key"] = 1

daily = (
    products
    .merge(dates, on="_key")
    .drop(columns="_key")
)

daily = daily.merge(
    all_data.groupby(
        ["product_id", "order_date"],
        as_index=False
    ).agg(
        sales_qty=("qty_invoiced", "sum"),
        price=("jeevee_variant_price", "mean")
    ),
    on=["product_id", "order_date"],
    how="left"
)

daily["sales_qty"] = daily["sales_qty"].fillna(0)

daily["price"] = (
    daily
    .sort_values(["product_id", "order_date"])
    .groupby("product_id")["price"]
    .ffill()
    .fillna(0)
)

daily = daily.sort_values(
    ["product_id", "order_date"]
).reset_index(drop=True)

```

```python
group = daily.groupby("product_id")["sales_qty"]

daily["lag_1"] = group.shift(1)
daily["lag_7"] = group.shift(7)
daily["lag_14"] = group.shift(14)
daily["lag_28"] = group.shift(28)



daily[daily['sales_qty']>0].sort_values(['sales_qty'])
```

```python
# Rolling Feature


daily["rolling_mean_7"] = (
    group.shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=0, drop=True)
)

daily["rolling_mean_14"] = (
    group.shift(1)
    .rolling(14)
    .mean()
    .reset_index(level=0, drop=True)
)

daily["rolling_mean_30"] = (
    group.shift(1)
    .rolling(30)
    .mean()
    .reset_index(level=0, drop=True)
)

daily["rolling_std_7"] = (
    group.shift(1)
    .rolling(7)
    .std()
    .reset_index(level=0, drop=True)
)

daily["rolling_std_30"] = (
    group.shift(1)
    .rolling(30)
    .std()
    .reset_index(level=0, drop=True)
)

daily.columns
```

```python
# Active Days
daily["active_day"] = (
    daily["sales_qty"] > 0
).astype(int)

active_group = daily.groupby("product_id")["active_day"]

daily["active_days_7"] = (
    active_group.shift(1)
    .rolling(7)
    .sum()
    .reset_index(level=0, drop=True)
)

daily["active_days_30"] = (
    active_group.shift(1)
    .rolling(30)
    .sum()
    .reset_index(level=0, drop=True)
)
```

```python
# ADU


daily["sales_7"] = (
    group.shift(1)
    .rolling(7)
    .sum()
    .reset_index(level=0, drop=True)
)

daily["sales_30"] = (
    group.shift(1)
    .rolling(30)
    .sum()
    .reset_index(level=0, drop=True)
)

daily["adu_7"] = (
    daily["sales_7"] /
    daily["active_days_7"].replace(0, np.nan)
)

daily["adu_30"] = (
    daily["sales_30"] /
    daily["active_days_30"].replace(0, np.nan)
)
```

```python
# Calender Feature


daily["day_of_week"] = (
    daily["order_date"].dt.dayofweek
)

daily["day_of_month"] = (
    daily["order_date"].dt.day
)

daily["month"] = (
    daily["order_date"].dt.month
)

daily["week_of_year"] = (
    daily["order_date"]
    .dt.isocalendar()
    .week
    .astype(int)
)

daily["is_weekend"] = (
    daily["day_of_week"] >= 5
).astype(int)
```

```python
# Target

daily["target"] = (
    daily
    .groupby("product_id")["sales_qty"]
    .shift(-1)
)
```

```python
# Feature list


features = [
    "lag_1",
    "lag_7",
    "lag_14",
    "lag_28",

    "rolling_mean_7",
    "rolling_mean_14",
    "rolling_mean_30",

    "rolling_std_7",
    "rolling_std_30",

    "active_days_7",
    "active_days_30",

    "sales_7",
    "sales_30",

    "adu_7",
    "adu_30",

    "price",

    "day_of_week",
    "day_of_month",
    "month",
    "week_of_year",
    "is_weekend"
]
```

```python
# Clean


daily = daily.replace(
    [np.inf, -np.inf],
    np.nan
)

daily[features] = (
    daily[features]
    .fillna(0)
)

```

```python
# split 2025 and 2026
train = daily[
    daily["order_date"] < "2026-01-01"
].copy()

test = daily[
    daily["order_date"] >= "2026-01-01"
].copy()
```

```python
print("Train:", train["order_date"].min(),
      "→", train["order_date"].max())

print("Test:", test["order_date"].min(),
      "→", test["order_date"].max())
```

```python
# don't use 2026 target during preduction
X_train = train[features]
y_train = train["target"]

X_test = test[features]
y_test = test["target"]
```

```python
# Train XGBOOST
model = XGBRegressor(
    n_estimators=500,
    max_depth=10,
    learning_rate=0.05,

    subsample=0.8,
    colsample_bytree=0.8,

    objective="reg:squarederror",

    random_state=42,
    n_jobs=-1
)

model.fit(
    X_train,
    y_train
)
```
